```python
import matplotlib.pyplot as plt
import numpy as np
import time
```

```python
def step_func (Vec_len):
    x = np.linspace(0, 1, Vec_len)  
    y = np.abs(np.where(np.sign(np.cos(2 * np.pi / 1 * x)) > 0, 1, 0)-1) # fliiping sides
    return y
    
def upwind(ai,aii,C):
    return ai - C*(ai-aii)

def main():
    C = 0.5
    u = 1 ; T = 1/u
    N = 64; dx = 1/N
    dt = C/u*dx
    step = []
    step.append(step_func(N))
    # print(step_func(N))
    print(step[0])
    step_0 = step[0][-2]
    for i in range(N):
        if i==0:
            step.append(upwind(step[0][-1],step_0,C))
        else:
            # print(step)
            step.append(upwind(step[-1][-1],step[-1][-2],C))
            
    plt.plot(step[-1])
    plt.show

if main:
    main()
        
    
    
```
