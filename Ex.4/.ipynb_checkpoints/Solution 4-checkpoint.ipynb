```python
def ghost_cell (phi):
    phi_ =  np.zeros(np.array(phi.shape)+2)
    phi_[1:-1,1:-1] = phi.copy()
    return phi_
    
def  Poisson_func_derichle (phi=[], f=0):
    Nx,Ny = phi.shape
    N = Nx
    phi_=ghost_cell (phi)
    phi_new = phi_*0
    for k in range(150):
        # phi_new = phi_*0
        for i in range(1,N+1):
            for j in range(1,N+1):
                phi_new[i,j] = 1/4 * (phi_[i+1,j] + phi_[i-1,j] + phi_[i,j+1] + phi_[i,j-1] + )
        phi = phi_new
    plt.imshow(phi_new)
    print(phi_new)
def main(N = 2**6):
    phi0 = np.ones([N,N]) # [raws,colmns]
    roh = 1 #normlized to epsilon_0
    Poisson_func_derichle (phi=phi0, f=roh)


if __name__ == "__main__":
    main()
    
    
    
    
```
